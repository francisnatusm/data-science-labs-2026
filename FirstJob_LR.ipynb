```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

np.random.seed(99)

#synthetic data for Years_Experience vs Annual_Salary
num_employees = 80
years_experience = np.random.rand(num_employees) * 15 + 1
noise_salary = np.random.randn(num_employees) * 5000
annual_salary = 40000 + years_experience * 3500 + noise_salary

salary_data = pd.DataFrame({'Years_Experience': years_experience, 'Annual_Salary': annual_salary})

display(salary_data.head())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Years_Experience', y='Annual_Salary', data=salary_data)
plt.title('Years of Eperience vs. Annual Salary (Synthetic Data)')
plt.xlabel('Years of Experience')
plt.ylabel('Annual Salary ($)')
plt.grid(True)
plt.show()
```

```python
X_salary = salary_data[['Years_Experience']]
y_salary = salary_data['Annual_Salary']

salary_model = LinearRegression()
salary_model.fit(X_salary, y_salary)

y_salary_pred = salary_model.predict(X_salary)

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Years_Experience', y='Annual_Salary', data=salary_data, label='Actual Salary')
plt.plot(X_salary, y_salary_pred, color='purple', linewidth=2, label='Refression Line')
plt.title('Linear Regression: Years of Experience vs. Annual Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Annual Salary ($)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
print(f"Intercept (Starting Salary with zero experience): {salary_model.intercept_:.2f}")
print(f"Coefficient for Years of Experience: {salary_model.coef_[0]:.2f}")

r_squared_salary = r2_score(y_salary, y_salary_pred)
print(f"R-squared: {r_squared_salary:.2f}")

mse_salary = mean_squared_error(y_salary, y_salary_pred)
print(f"Mean Squared Error: {mse_salary:.2f}")
```
